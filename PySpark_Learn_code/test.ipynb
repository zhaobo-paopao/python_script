```python
from collections import defaultdict
d=defaultdict(list)
d['hello'].append(1)
d
```

```python
from collections import defaultdict

dd = defaultdict(list)
print("第一次访问 key='x' 之前，字典内容：", dict(dd))
dd['x'].append(100)
print("访问并 append 之后，字典内容：", dict(dd))
print("dd['x'] 的值是：", dd['x'])
```

```python
list(map(lambda x: x*2, [1, 2, 3]))  # [2, 4, 6]
```

```python
import itertools
nested = [[1, 2], [3, 4]]
list(itertools.chain.from_iterable(nested))  # [1, 2, 3, 4]
```

# Series的创建

```python
# Series的创建通过列表来创建
import pandas as pd
s=pd.Series([10,2,3,4,5])
s=pd.Series([10,2,3,4,5],index=['A','B','C','D','E'])  #自定义索引
# s=pd.Series([1,2,3,4,5],index=('A','B','C','D','E'))
s=pd.Series([10,2,3,4,5],['A','B','C','D','E'],name='月份') #定义name

print(s)
```

```python
# 通过字典来创建Series
s1=pd.Series({'a':'1','b':'2','c':'3','d':'4','e':'5'})
s2=pd.Series(s1,index=['a','b'])  #s2通过取s1的 a 和 b 索引来创建
print(s1)
print(s2)
```

```python
#Series的属性
s=pd.Series([1,2,3,4,5],index=['a','b','c','d','e'])  #自定义索引
s.name='test'
print(s)
print('索引:',s.index) #索引
print('获取值:',s.values) #获取值
print('获取形状(长度):',s.shape) #Series是一维的,所以只有长度
print('获取维度:',s.ndim)
print('获取元素的个数:',s.size)
print('获取Series的类型:',s.dtype)
print('获取Series的名字:',s.name)
print('显式索引a的值为:',s.loc['a'])  #根据索引来获取它的值(显式索引)
print('隐式索引a的值为:',s.iloc[0])  #根据索引来获取它的值(隐式索引)
print('显式索引切片:',s.loc['a':'c'])  #根据索引来获取它的值(显式索引)
print('隐式所以切片为:',s.iloc[0:3])  #根据索引来获取它的值(隐式索引)
print('获取单个元素a的值:',s.at['a']) #专门用于快速获取单个标量值
# print(s.a)  #输出1
# print(s['a']) #输出1
# print(s.at['a']) #输出1
# print(s.loc['a']) #输出1


```

```python
# Series学生成绩统计
# 创建一个包含10名学生数学成绩的Series,成绩范围50-100之间,计算平均分，最高分，最低分，并找出高于平均分的学生人数
np.random.seed(42)
scores=pd.Series(np.random.randint(50,101,10),index=['学生'+str(i) for i in range(1,11)])
print(scores)
```

```python
count=(scores>scores.mean())
above_score=scores[scores>scores.mean()]
print('学生数学成绩平均分:',scores.mean())
# print('学生数学成绩平均分:',np.mean(scores))
print('学生数学成绩最高分:',scores.max())
print('学生数学成绩最低分:',scores.min())
print('高于平均分的学生人数:',count.sum())
print('高于平均分的学生和成绩为:',above_score)

```

```python
#温度数据分析
# 给定某城市一周每天的最高温度Series,完成以下任务
# 找出温度大于30度的天数
# 计算平均温度
# 将温度从高到低排序
# 找出温度变化最大的两天
temperatures=pd.Series([28,31,29,32,30,27,33],index=['周一','周二','周三','周四','周五','周六','周日'])
print('温度大于30度的天:',temperatures[temperatures>30])
print('温度大于30度的天数:',temperatures[temperatures>30].count())
print('平均温度:',temperatures.mean())
# temperatures.sort_values(ascending=False) #温度从高到低排序
print('温度从高到低排序:',temperatures.sort_values(ascending=False))
# print(f'最大值:{temperatures.idxmax()}={temperatures.max()}度')
# print(f'最小值:{temperatures.idxmin()}={temperatures.min()}度')
# t1=temperatures.max()
# t2=temperatures.min()
# t3=t1-t2
# print(f'温度最高的是:{temperatures.idxmax()}',f'温度是:{temperatures.max()}度',f'温度最低的是:{temperatures.idxmin()}',f'温度是:{temperatures.min()}度',f'温度相差{t3}度')
# print(te1)
t1=temperatures.diff().abs() #计算Series变化的值
# t1.max()
print('温度变化最大的两天是:',t1.sort_values(ascending=False).index[0:2].to_list())
# t1.sort_values(ascending=False).keys()
# print('温度变化最大的两天是:',t1.sort_values(ascending=False))

```

```python
#销售数据分析
# 某产品过去12个月的销售Series
# 计算季度平均销量(每三个月为一个季度)
# 找出销量最高的月份
# 计算月环比增长率
# 找出连续增长超过2个月的月份
sales=pd.Series([120,135,145,160,155,170,180,175,190,200,210,220],index=pd.date_range('2022-01-01',periods=12,freq='MS'))
print(sales)
```

```python
#季度平均销量
sales.resample('QS').mean() #'Q'按季度重新采样求平均值
# sales.resample('QS').sum() #'QS'按季度重新采样求和
# sales.resample('YS').mean() #'YS'按年求平均值
print('销量最高的销售额:',sales.max(),'销量最高的月份:',sales.idxmax())
```

```python
#统计月环比的增长率
print('月环比的增长率:\n',sales.pct_change())
```

```python
# 连续增长超过2个月的月份
a=sales.pct_change()
b=a>0
b[b.rolling(3).sum()==3].keys().tolist()
```

```python
#没小时数据分析
# 某个商店每小时销售额Series
# 按天重新采集每日总销售额
# 计算每天营业时间8-22点和非营业时间的销售额比例
# 找出销售额最高的三个小时
np.random.seed(42)
hourly_sales=pd.Series(np.random.randint(0,100,24),index=pd.date_range('2025-01-01',periods=24,freq='H'))
print(hourly_sales)
```

```python
#按天进行重采样,一天24个小时进行汇总
day_sales=hourly_sales.resample('D').sum()  #一天总营业额
#取部分时间的数据
# hourly_sales.between_time('8:00','22:00')

business_hours_sales= hourly_sales[(hourly_sales.index.hour>=8) & ((hourly_sales.index.hour<=22))].sum() #营业时间营业额
# not_business_hours_sales=hourly_sales.index.drop(business_hours_sales.index) #非营业时间营业额
print('营业时间8-22点和非营业时间的销售额比例:\n',business_hours_sales/(day_sales-business_hours_sales))  #一天总营业额-营业时间营业额=非营业时间营业额
print('销售额最高的三个小时:\n',hourly_sales.nlargest(3))#取销售额最高的三个小时
```

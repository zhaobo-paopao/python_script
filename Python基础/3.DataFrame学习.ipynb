```python
# 练习2:销售数据分析
# 某公司销售数据如下
# 1：计算每种产品的总销售额=单价*数量
# 2：找出销售额最高的产品
# 3：按照销售额由高到低排列，并输出所有产品信息
data={
      '产品名称':['A','B','C','D'],
      '单价':[100,150,200,120],
      '销量':[50,30,20,40]
      }
sales=pd.DataFrame(data)
# 计算每种产品的总销售额=单价*数量
print('计算每种产品的总销售额=单价*数量:')
sales['总销售额']=sales['单价']*sales['销量']
# 找出销售额最高的产品
sales.nlargest(1,columns='总销售额')
#按照销售额由高到低排列，并输出所有产品信息
sales.sort_values('总销售额',ascending=False)



```

```python
# 1.计算每位用户的总消费金额(消费金额=商品单价x购买数量)
# 2.找出消费金额最高的用户，并输出其所有信息
# 3.计算所有用户的平均消费金额(保留2位小数)
# 4.统计电子产品的总购买数量
data1={
    '用户ID':[101,102,103,104,105],
    '用户名':['Alice','Bob','Charlie','David','Eve'],
    '商品类别':['电子产品','服饰','电子产品','家居','服饰'],
    '商品单价':[1200,300,800,150,200],
    '购买数量':[1,3,2,5,4]
}
df=pd.DataFrame(data1)
# 计算每位用户的总消费金额(消费金额=商品单价x购买数量)
df['总消费金额']=df['商品单价']*df['购买数量']
# 找出消费金额最高的用户，并输出其所有信息
df.nlargest(1,columns='总消费金额')
# 计算所有用户的平均消费金额(保留2位小数)
df['总消费金额'].mean()
# 统计电子产品的总购买数量
df[df['商品类别']=='电子产品']['购买数量'].sum()

# df
```

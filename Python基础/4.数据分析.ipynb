```python
#数据采样中分组聚合
# df.groupby('分组字段')['聚合字段'].聚合函数()

df=pd.read_csv('data/employee.csv')
# df.info()
df.dropna(subset='salary')
df['salary']=df['salary'].astype('Int64')
# print(df['salary'].isna().sum())
#计算不同部门的平均工资保留两位小数
df.groupby('department')['salary'].mean().round(2).sort_values(ascending=False)
# df.groupby('department').groups  #查看分组
# df.groupby('department').get_group('技术部')  #查看具体的每个分组
# df2.sort_values('salary')
```

```python
#计算不同部门不同工作年限的人的平均工资
df2=df.groupby(['department','work_years'])['salary'].mean()
df2=df2.reset_index()
df2.sort_values('salary',ascending=False)  #按照薪水求平均数,然后再降序排列
```

```python
#企鹅数据分析
# 1：导入必要的库
import pandas as  pd
import numpy  as np
# 2：导入数据
df=pd.read_csv('data/penguins.csv')
df.info()
```

```python
# 3：数据清洗
# 缺失值检查
print(df.isna().sum())
#删除缺失值,inplace=True立即生效
df.dropna(inplace=True) 
df.isna().sum()
len(df)
```

```python
# 4：数据特征构造
# df.head()
# 将性别转化为类别类型
df['sex']=df['sex'].astype('category')
df['bill_ratio']=df['bill_length_mm']/df['bill_depth_mm']
# df.info()
df.head()
```

```python
# 5：数据分析
# 数据分箱,把体重分为低中高三个等级
labels=['低','中','高']
df['mess_level']=pd.cut(df['body_mass_g'],bins=3,labels=labels)
# df = df.drop('体重等级', axis=1) #删除某列
print(df['mess_level'].value_counts())
#按照岛屿和性别进行分组
df2=df.groupby(['island','sex']).agg({
    'body_mass_g':['mean','count']
}

)
df2=df2.reset_index()
df2    

```

```python
#睡眠数据分析
# 1：导入库
import pandas as pd
import numpy as np
# 2:导入数据
df=pd.read_json('data/sleep_data.json')
df.head()
# df.info()
# df.describe()
df.dropna(inplace=True)
df.isna().sum()

```

```python
# 3:数据清洗
# df.isna().sum()
df['sleep_duration'].value_counts()
# df['deep_sleep'].value_counts()
# df['light_sleep'].value_counts()
# df['wake_times'].value_counts()
# df['sleep_score'].value_counts()
# df['heart_rate_avg'].value_counts()
# df['sleep_quality'].value_counts()
df['sleep_duration']=df['sleep_duration'].astype('category')
df.info()
#数据分箱
labels=['差','中','优']
df['sleep_quality']=pd.cut(df['deep_sleep'],bins=3,labels=labels)
sleep_labels=['没睡好','睡的还行','睡的很好']
df['睡眠质量']=pd.cut(df['sleep_score'],bins=3,labels=sleep_labels)
df.head()
```

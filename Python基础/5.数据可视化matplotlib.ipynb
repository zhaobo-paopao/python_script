```python
#绘制折线图
import matplotlib.pyplot as plt
from matplotlib import rcParams #导入这个库是控制中文字体格式
rcParams['font.family']='SimHei'  #windows系统 'SimHei'设置中文字体
#创建图表,设置大小
plt.figure(figsize=(10,5))   #宽是10高是5
#绘图的数据
month=['1月','2月','3月','4月']
sales=[100,150,80,130]
#绘制折线图
plt.plot(month,sales,
         label='产品A',
         color='blue',
         linewidth=2,#linewidth=2设置折线的粗细
         linestyle='--',
         marker='o')  # marker='o设置折现点的圆点
#添加标题
plt.title('2025年前四个月销售趋势',color='red',fontsize=15)
#添加坐标轴的标签
plt.xlabel('月份',fontsize=10)
plt.ylabel('销售额(万元)',fontsize=10)
#添加图例
plt.legend(loc='upper left')
#添加网格线
# plt.grid(True,axis='y') #添加X或者Y轴的网格线
plt.grid(True,alpha=0.1,color='red',linestyle='--')#默认X和Y都添加网格线,并设置透明度和颜色，网格线的风格
#设置刻度字体大小
# plt.xticks(rotation=60,fontsize=10)#rotation=0代表字体的旋转角度,当X和Y轴的字体过多的时候可以设置这个参数控制旋转角度
plt.xticks(rotation=0,fontsize=10)#rotation=0代表字体的旋转角度,当X和Y轴的字体过多的时候可以设置这个参数控制旋转角度
#设置y轴的范围
# plt.xticks()
plt.ylim(0,160) #默认是80开始,可以设置从0开始到160的范围.
#在每个数据点显示数值
for x,y in zip(month,sales):
    # print(x,y)
    plt.text(x,y+3,
    str(y),ha='center',va='bottom',fontsize=10) #y+3控制数字和折现点的距离

#显示图标
plt.show()
```

```python
#绘制散点图，查看数据分布以及离散程度
import matplotlib.pyplot as plt
from matplotlib import rcParams #导入这个库是控制中文字体格式
import random 
rcParams['font.family']='SimHei'  #windows系统 'SimHei'设置中文字体
#创建图表,设置大小
plt.figure(figsize=(10,8))   #宽是10高是5
#绘图的数据
x=[]
y=[]
for i in range(1000):
    tmp=random.uniform(0,10)
    x.append(tmp)
    tmp2=2*tmp+random.gauss(0,2)
    # print(tmp,tmp2)
    y.append(tmp2)
# # color=[] #配色方案
# #绘制折线图
plt.scatter(x,y,
            color='blue',  #颜色
            alpha=0.5,  #透明度
            s=20,  #圆点大小
            label='数据'
            )
# #添加标题
plt.title('x变量与y变量的关系',color='red',fontsize=15)
#添加坐标轴标签
plt.xlabel('自变量X',fontsize=10)
plt.ylabel('因变量Y',fontsize=10)
#添加图例
plt.legend(loc='upper left')
#添加网格线
plt.grid(True,alpha=0.1,color='blue',linestyle='--')
#设置刻度字体大小
plt.xticks(rotation=0,fontsize=12)
plt.yticks(rotation=0,fontsize=12)
#设置y轴的范围
plt.ylim(0,30)
plt.plot([0,10],[0,20],color='red',linewidth=2,linestyle='--')
# plt.pie(times,
#         labels=things,
#         autopct='%1.1f%%', #显示百分比
#         startangle=90, #逆时针旋转90度
#         # colors=color/*
#         wedgeprops={'width':0.7}, #设置圆环百分比
#         pctdistance=0.6   #设置百分比的距离
# )
# plt.text(0,.0,'总计:\n100%',ha='center',va='bottom',fontsize=10)



plt.tight_layout()
#显示图标
plt.show()
```

```python
#案例分析
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import rcParams
plt.rcParams['font.sans-serif'] = ['SimHei', 'DejaVu Sans']  # 设置中文字体
plt.rcParams['axes.unicode_minus'] = False  # 关闭unicode负号，使用ASCII负号
plt.rcParams['font.size'] = 12  # 设置默认字体大小
# rcParams['font.family']='SimHei'
df=pd.read_csv('data/xian_weather.csv')

# 绘制气温趋势变化折线图
df['date']=pd.to_datetime(df['date'])
# df.head()
plt.figure(figsize=(15,10))
plt.plot(df['date'],df['high_temp'],label='最高气温')
plt.plot(df['date'],df['low_temp'],label='最低气温')
plt.title('2025年西安气温变化图')
plt.xlabel('日期')
plt.ylabel('温度℃')
plt.legend(loc='upper left')
df.head()
```

```python
df['mean_temp']=(df['high_temp']+df['low_temp'])/2
plt.figure(figsize=(15,10))
plt.plot(df['date'],df['mean_temp'],label='平均气温')
# plt.plot(df['date'],df['low_temp'],label='最低气温')
plt.title('2025年西安平均气温变化图')
plt.xlabel('日期')
plt.ylabel('温度℃')
plt.legend(loc='upper left')
df.head()
```
